```python
orders = {"Luisa Bauer": (7.5, 32.00), "Ferdi Fuchs": (5.0, 27.42), 
          "Erika Mustermann": (2.2, 23.27), "Max Müller": (4.3, 14.95)}

#ENTER YOUR CODE BELOW
def get_final_cost(orders):
    final_costs = {}
    for name, (dist, price) in orders.items():
        if dist <= 3.0:
            delivery = 0.0
        elif dist <= 5.0:
            if price < 20:
                delivery = 5.0
            else:
                delivery = 0.0
        else:
            delivery = (dist * 10) * 0.10

        final_costs[name] = round(price + delivery, 2)

    return final_costs

#END OF YOUR CODE

print(get_final_cost(orders))  # a dictionary has to be printed here!
```

```python
# define the lists 'grades' and 'students' and find the average grade for the course
grades = [ 2.3, 1.7, 2.7, 3.3, 2.0, 4.7 ]
students = [ "Alfred", "Berta", "Charlotte", "Daniel", "Eva", "Felix" ]

average_grade = 0.0
for i in range(6):
    average_grade += grades[i] 
average_grade /= 5

print("The average grade in the course is " + str(average_grade) + ".")
```

```python
# create a dictionary called 'gradebook' using the values of 'grades' and 'students'
gradebook = {}
for i in range(len(grades)):
    gradebook[students[i]] = grades[i]
print(gradebook)
```

```python
# check if 'Felix' failed the exam (grade worse than 4.0); if so, change his grade to the string "Failed"
# NOTE: if this cell fails unexpectedly due to a type error, execute the answer cell above it again
gradebook_new = gradebook.copy()
if (gradebook_new["Felix"] > 4.0):
    gradebook_new["Felix"] = "Failed"
print(gradebook_new)
```

```python
# create a new dictionary with all lower case keys and the same values as before
# HINT: Look up a list of methods that can be used on strings
gradebook_lower = {}
for x in gradebook_new.items():
    gradebook_lower[x[0].lower()] = x[1]
print(gradebook_lower)
```

```python
"""Read only helper cell - not editable"""
# given a,b,c and an array of x-values, this function calculates a*x^2 + b*x + c
def foo(a, b, c, x):
    for val in x:
        d = a*(val**2) + b*val + c
        print(f"{a}*{val}^2 + {b}*{val} + {c} = {d}")
    return d
```

```python
"""Read only helper cell - not editable"""
def bar(x, y): 
    if x == 0: 
        return y 
    elif x == 4: # skip x = 4 to avoid division by zero 
        return bar(x - 1, y) 
    else: temp = 2 / (x - 4) 
    return bar(x - 1, y + temp)
```
